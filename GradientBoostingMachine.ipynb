# Modelo de Classificaçao do GBM

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.datasets import make_hastie_10_2
from sklearn.model_selection import GridSearchCV
```

```python
X, y = make_hastie_10_2(random_state=42)
X_train, X_test = X[:2000], X[2000:]
y_train, y_test = y[:2000], y[2000:]

clf = HistGradientBoostingClassifier(max_iter=100, random_state=42).fit(X_train, y_train)
clf.score(X_test, y_test)
```

```python
# Parametros para o Grid search
param_grid = {
    'max_iter': [180, 200, 220],  # Número de iterações (equivale a n_estimators)
    'learning_rate': [0.9, 1.0, 1.1],  # Taxa de aprendizado
    'max_depth': [1, 2, 3],  # Profundidade máxima das árvores
    'min_samples_leaf': [15, 20, 25],  # Mínimo de amostras em uma folha
    'l2_regularization': [0.0, 0.1, 1.5]  # Regularização L2 para evitar overfitting
}
```

```python
# busca com validação cruzada
grid_search = GridSearchCV(clf, param_grid, cv=5, scoring='accuracy', n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)
```

```python
# Exibir os melhores hiperparâmetros encontrados
print("Melhores hiperparâmetros:", grid_search.best_params_)
```

```python
# Avaliar a performance do melhor modelo encontrado
best_model = grid_search.best_estimator_
accuracy = best_model.score(X_test, y_test)
print(f"Acurácia no conjunto de teste: {accuracy:.4f}")
```

```python
# 1 Tunning
param_grid = {
    'max_iter': [210, 220, 230],  # Número de iterações (equivale a n_estimators)
    'learning_rate': [1.0],  # Taxa de aprendizado
    'max_depth': [1],  # Profundidade máxima das árvores
    'min_samples_leaf': [12, 15, 17],  # Mínimo de amostras em uma folha
    'l2_regularization': [0.1, 1.2]  # Regularização L2 para evitar overfitting
}
# busca com validação cruzada
grid_search = GridSearchCV(clf, param_grid, cv=5, scoring='accuracy', n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)

# Exibir os melhores hiperparâmetros encontrados
print("Melhores hiperparâmetros:", grid_search.best_params_)
```

```python
# 2 Tunning
param_grid = {
    'max_iter': [205, 210, 215],  # Número de iterações (equivale a n_estimators)
    'learning_rate': [1.0],  # Taxa de aprendizado
    'max_depth': [1],  # Profundidade máxima das árvores
    'min_samples_leaf': [10, 12, 13],  # Mínimo de amostras em uma folha
    'l2_regularization': [0.1]  # Regularização L2 para evitar overfitting
}
# busca com validação cruzada
grid_search = GridSearchCV(clf, param_grid, cv=5, scoring='accuracy', n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)

# Exibir os melhores hiperparâmetros encontrados
print("Melhores hiperparâmetros:", grid_search.best_params_)
```

```python
# Avaliar a performance do melhor modelo encontrado
best_model = grid_search.best_estimator_
accuracy = best_model.score(X_test, y_test)
print(f"Acurácia no conjunto de teste: {accuracy:.4f}")
```

# Modelo de Regressão do GBM

```python
from sklearn.datasets import make_regression
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import train_test_split, GridSearchCV
import numpy as np
```

```python
X, y = make_regression(random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

reg = GradientBoostingRegressor(random_state=42)
reg.fit(X_train, y_train)
```

```python
reg.predict(X_test[1:2])
```

```python
reg.score(X_test, y_test)
```

```python
# Parametros para Gridsearch
param_grid = {
    'n_estimators': [200, 210, 220],  # Número de iterações (equivale a n_estimators)
    'learning_rate': [0.15, 0.2, 0.3],  # Taxa de aprendizado
    'max_depth': [3, 5, 7],  # Profundidade máxima das árvores
    'min_samples_leaf': [2, 3, 4],  # Mínimo de amostras em uma folha
    'min_samples_split': [2, 5, 10] # Minimo de amostras para dividir um nó
}
```

```python
#Validação cruzada
grid_search = GridSearchCV(reg, param_grid, cv=5, scoring='neg_mean_squared_error', n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)
```

```python
print("Melhores Hiperparametros:", grid_search.best_params_)
```

```python
#Tunning hiperparametros
param_grid = {
    'n_estimators': [215, 220, 225],  # Número de iterações (equivale a n_estimators)
    'learning_rate': [0.15],  # Taxa de aprendizado
    'max_depth': [2, 3, 4],  # Profundidade máxima das árvores
    'min_samples_leaf': [4, 5, 6],  # Mínimo de amostras em uma folha
    'min_samples_split': [8, 10, 12] # Minimo de amostras para dividir um nó
}

#Validação cruzada
grid_search = GridSearchCV(reg, param_grid, cv=5, scoring='neg_mean_squared_error', n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)

print("Melhores Hiperparametros:", grid_search.best_params_)
```

```python
from sklearn.model_selection import cross_val_score

best_model_reg = grid_search.best_estimator_
cv_scores = cross_val_score(best_model_reg, X_test, y_test, cv=5, scoring="neg_mean_squared_error")

print(f"Média do MSE na Validação Cruzada: {np.mean(-cv_scores):.2f}")
```

```python
from sklearn.metrics import mean_squared_error

# Calculando o MSE para treino e teste
mse_train = mean_squared_error(y_train, best_model_reg.predict(X_train))
mse_test = mean_squared_error(y_test, best_model_reg.predict(X_test))

print(f"MSE Treino: {mse_train:.2f}")
print(f"MSE Teste: {mse_test:.2f}")

```

```python
#Tunning 2
param_grid = {
    'n_estimators': [600],  # Número de iterações (equivale a n_estimators)
    'learning_rate': [0.0001],  # Taxa de aprendizado
    'max_depth': [2],  # Profundidade máxima das árvores
    'min_samples_leaf': [5],  # Mínimo de amostras em uma folha
    'min_samples_split': [12] # Minimo de amostras para dividir um nó
}

#Validação cruzada
grid_search = GridSearchCV(reg, param_grid, cv=5, scoring='neg_mean_squared_error', n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)

print("Melhores Hiperparametros:", grid_search.best_params_)
```
