# Tarefa - Álgebra Linear

A célula abaixo gera dados de notas de duas provas, p1 e p2, conforme em aula, mas com mais alunos.

## 1. Reduzindo correlações

a) Já vimos que expressar os dados através da média e diferença é uma transformação linear. Obtenha a média e a diferença através de uma multiplicação matricial, conforme visto em aula.

b) Calcule a correlação entre os dados da **p1** e **p2**

c) Calcule a correlação entre os dados transformados (média e diferença)

d) Comente as correlações observadas.

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

# Gerando os dados
np.random.seed(123)

p1 = np.random.random(50) *10
p1[p1>10] = 10
p1[p1<0] = 0

p2 = p1 + np.random.normal(0,1,50)
p2[p2>10] = 10
p2[p2<0] = 0

df = pd.DataFrame({'p1': p1, 'p2': p2})
```

```python
df['média'] = df.mean(axis=1)
df['diferença'] = df.p2 - df.p1
df.style.format(precision=1, decimal=',')
```

```python
# Correlação entre os dados
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Gráfico original (p1 vs p2)
sns.scatterplot(x=df['p1'], y=df['p2'], ax=ax[0])
ax[0].set_title('Correlação Original: p1 vs p2')

# Gráfico transformado (média vs diferença)
sns.scatterplot(x=df['média'], y=df['diferença'], ax=ax[1])
ax[1].set_title('Correlação Transformada: Média vs Diferença')

plt.show()
```

**Correlação Original: p1 vs p2**
- Mostra uma forte correlaçao linear, que indica que os valores eram parecidos

**Correlação Transformada: Média vs Diferença**
- Mostra pontos espalhados sem correlação linear visivel, pois a transformação removeu a dependência linear entre p1 e p2.


## 2. Soma e diferença

Realize agora uma transformação semelhante, mas que ao invés da média, dê a soma da **p1** e **p2**. Construa o gráfico e conclua que diferenças visuais você observa.

```python
df['soma'] = df.p1 + df.p2
```

```python
# Correlação entre os dados
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Gráfico original (p1 vs p2)
sns.scatterplot(x=df['p1'], y=df['p2'], ax=ax[0])
ax[0].set_title('Correlação Original: p1 vs p2')

# Gráfico transformado (média vs diferença)
sns.scatterplot(x=df['soma'], y=df['diferença'], ax=ax[1])
ax[1].set_title('Correlação Transformada: Soma vs Diferença')

plt.show()
```

O grafico da correlacao da Soma vs Diferença parece igual o grafico da Média vs Diferença. Isso acontece porque a transformação da soma e a media são linearmente dependentes e assim os pontos no gráfico apenas mudam de escala, mas a forma da distribuição dos pontos continua igual.
