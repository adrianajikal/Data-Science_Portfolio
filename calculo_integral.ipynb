1- Calcule de forma numérica a integral da função $x^3$ avaliada entre os pontos $1$ e $2$. Identifique o número de retângulos que você usou.  
2- Calcule essa área de forma analítica.

```python
import numpy as np
```

```python
f = lambda x: x**3

def calcula_area(a, b, func, frame_num):
    num_bins = frame_num+1 #quantidade de retangulos
    delta = (b-a)/num_bins #largura do retangulo
    x_cols = np.linspace(a, b-delta, num_bins) + delta/2 #divisao do eixo X
    y_cols = f(x_cols) #altura dos retangulos
    return sum(delta * y_cols)

a = 1 
b = 2
num_retangulos = 1000000 #numero de retangulos

area_aproximada = calcula_area(a, b, f, num_retangulos)
print(f"Area aproximada: {area_aproximada}")
print(f"Numero de retangulos utilizados: {num_retangulos}")
```
